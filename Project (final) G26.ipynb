### Phương pháp
##### 1. Thuật toán tìm kiếm được áp dụng
- **DFS (Depth First Search – Tìm kiếm theo chiều sâu)**
    - Cách hoạt động: đi sâu vào một nhánh cho đến khi không thể tiếp tục, sau đó quay lui (backtrack)
    - Ưu điểm: đơn giản, ít tốn bộ nhớ
    - Nhược điểm: không đảm bảo tìm được đường đi ngắn nhất, có thể duyệt nhiều nhánh thừa
- **BFS (Breadth First Search – Tìm kiếm theo chiều rộng)**
    - Cách hoạt động: duyệt từng lớp từ điểm xuất phát, mở rộng đồng đều theo tất cả các hướng.
    - Ưu điểm: đảm bảo tìm được đường đi ngắn nhất (nếu tồn tại)
    - Nhược điểm: tốn bộ nhớ khi mê cung lớn do lưu trữ nhiều nút trung gian
- **A* (A-star Search)**
    - Cách hoạt động: Mỗi ô trong mê cung được gán giá trị:   f(n) = g(n) + h(n)
    Trong đó:  
        - `g(n)`: chi phí từ điểm bắt đầu đến ô hiện tại  
        - `h(n)`: ước lượng chi phí từ ô hiện tại đến đích (heuristic)  
        - Tại mỗi bước, thuật toán chọn ô có `f(n)` nhỏ nhất để mở rộng.  
        - Khi đến đích, A\* đảm bảo tìm ra đường đi **ngắn nhất**.
    - Ưu điểm: nhanh hơn BFS/Dijkstra trong mê cung lớn, đồng thời vẫn đảm bảo tìm được đường đi tối ưu.
    - Nhược điểm: cần chọn heuristic phù hợp và tiêu tốn nhiều bộ nhớ cho các trạng thái mở rộng.

### Chương trình
##### 1. Xây dựng thuật toán duyệt theo chiều sâu (DFS)

```python
def dfs_on_graph(graph, start, goal):
    #Khởi tạo ban đầu
    stack = [start]
    came_from = {start: None}
    yield ("visit", start)
    #Vòng lặp duyệt đồ thị, duyệt các đỉnh kề
    while stack:
        cur = stack.pop()
        yield ("expand", cur)
        if cur == goal: 
            break
        for (nbr, _) in graph.neighbors(cur):
            if nbr not in came_from:
                came_from[nbr] = cur
                stack.append(nbr)
                yield ("visit", nbr)
```

##### 2. Xây dựng thuật toán duyệt theo chiều rộng (BFS)

```python
def bfs_on_graph(graph, start, goal):
    #Khởi tạo ban đầu
    q = deque([start])
    came_from = {start: None}
    yield ("visit", start)
    #Vòng lặp chính, duyệt các nút kề
    while q:
        cur = q.popleft()
        yield ("expand", cur)
        if cur == goal:
            break
        for (nbr, _) in graph.neighbors(cur):
            if nbr not in came_from:
                came_from[nbr] = cur
                q.append(nbr)
                yield ("visit", nbr)
```

##### 3. Xây dựng thuật toán Dijkstra

```python
def dijkstra_on_graph(graph, start, goal):
    #Khởi tạo ban đầu
    heap = [(0, start)]
    dist = {start: 0}
    came_from = {start: None}
    yield ("visit", start)
    #Vòng lặp chính, duyệt tất cả hàng xóm 
    while heap:
        d, cur = heapq.heappop(heap)
        yield ("expand", cur)
        if cur == goal:
            break
        for (nbr, w) in graph.neighbors(cur):
            nd = d + w
            if nbr not in dist or nd < dist[nbr]:
                dist[nbr] = nd
                came_from[nbr] = cur
                heapq.heappush(heap, (nd, nbr))
                yield ("visit", nbr)
```

##### 4. Xây dựng thuật toán tìm đường đi ngắn nhất có hàm heuristic (A*)

```python
def astar_on_graph(graph, start, goal):
    #Hàm heuristic (ước lượng chi phí còn lại) giữa hai ô a và b.
    def h(a, b):
        return abs(a[0]-b[0]) + abs(a[1]-b[1])
    #Khởi tạo ban đầu
    heap = [(h(start, goal), 0, start)]
    gscore = {start: 0}
    came_from = {start: None}
    yield ("visit", start)
    #Vòng lặp chính, duyệt các láng giềng 
    while heap:
        _, curg, cur = heapq.heappop(heap)
        yield ("expand", cur)
        if cur == goal:
            break
        for (nbr, w) in graph.neighbors(cur):
            tentative_g = gscore[cur] + w
            if nbr not in gscore or tentative_g < gscore[nbr]:
                gscore[nbr] = tentative_g
                f = tentative_g + h(nbr, goal)
                came_from[nbr] = cur
                heapq.heappush(heap, (f, tentative_g, nbr))
                yield ("visit", nbr)
```

```python
def reconstruct_path(came_from, goal):
    path = []
    cur = goal
    while cur is not None:
        path.append(cur)
        cur = came_from[cur]
    path.reverse()
    return path
```

#### Các thuật toán giải quyết ma trận
##### 1. Tìm kiếm theo chiều rộng - Breadth-First Search (BFS)
##### a. Phương pháp:
BFS khám phá đồ thị theo tầng (level) từ nút bắt đầu.
- Dùng queue (FIFO) để mở rộng các nút theo tầng (level).
- Mỗi lần lấy một nút từ queue, đánh dấu là đã mở rộng.
- Thêm các neighbor chưa thăm vào queue.
- Khi đến goal, dừng và tạo đường đi từ came_from.

Điều này đảm bảo tìm được đường đi ngắn nhất trong đồ thị không trọng số.
##### b. Dữ liệu cần: 
- queue (sử dụng collections.deque) để lưu các node sắp thăm theo thứ tự FIFO.
- came_from lưu node cha của mỗi node, dùng để dựng lại đường đi cuối cùng.
- Graph phải có phương thức neighbors(node) trả về danh sách các node kề. Không cần trọng số các cạnh
##### c. Cách cài đặt
- Thêm node start vào queue `q = deque([start])`, `deque` đảm bảo node vào trước ra trước (FIFO), lưu trong came_from là None (`came_from = {start: None}`), `came_from` ghi lại cây BFS. Đưa start vào hàng đợi `yield ("visit", start)`
- Dùng vòng lặp while queue để xử lý:
    - Lấy node được thêm sớm nhất ra (popleft) (`cur = q.popleft()`) và đánh dấu expand (`yield ("expand", cur)`).
    - Nếu node đó là goal, dừng vòng lặp.
    - Duyệt toàn bộ neighbor, bỏ qua trọng số, `if nbr not in came_from` điều kiện đảm bảo không thăm lại node cũ tránh vòng lặp vô hạn
    - Với từng neighbor chưa thăm,  lưu vết đường đi `came_from[nbr] = cur`, thêm vào queue (`q.append(nbr)`), đánh dấu visit (`yield ("visit", nbr)`).
- Khi kết thúc, dùng reconstruct_path(came_from, goal) để dựng đường đi từ start → goal. 
##### d. Đặc điểm
- Tìm đường đi ngắn nhất trong đồ thị không trọng số
- Thời gian: O(V + E) 
##### 2. Tìm kiếm theo chiều sâu - Depth-First Search (DFS)
##### a. Phương pháp
DFS khám phá theo nhánh sâu trước, dùng satck (LIFO) để đi sâu đến tận cùng một nhánh trước khi quay lui. 
- Mỗi nút chỉ thăm một lần, neighbor chưa thăm mới thêm vào stack.
- Khi đến goal, dừng và dựng đường đi từ came_from.

Nó không đảm bảo đường đi ngắn nhất nhưng hiệu quả khi cần tìm tất cả các đường đi hoặc kiểm tra liên thông.
##### b. Dữ liệu cần 
- stack (list) để lưu các node chưa thăm theo thứ tự LIFO.
- came_from lưu node cha của mỗi node, dùng để dựng lại đường đi cuối cùng.
- Graph phải có phương thức neighbors(node). Không cần trọng số các cạnh
##### c. Cách cài đặt
- Thêm node start vào stack (`stack = [start]`), stack vào sau ra trước (LIFO), lưu trong came_from là None (`came_from = {start: None}`). Đưa start vào hàng đợi `yield ("visit", start)`.
- Dùng vòng lặp while stack:
    - Lấy node được thêm vào muộn nhất ra (stack.pop()) và đánh dấu expand.
    - Nếu node là goal, dừng vòng lặp.
    - Duyệt toàn bộ neighbor, bỏ qua trọng số, `if nbr not in came_from` điều kiện đảm bảo không thăm lại node cũ tránh vòng lặp vô hạn.
    - Với từng neighbor chưa thăm, lưu vết đường đi `came_from[nbr] = cur`, thêm vào stack (`stack.append(nbr)`), đánh dấu visit (`yield ("visit", nbr)`).
- Khi kết thúc, dựng đường đi bằng reconstruct_path(came_from, goal).
##### d. Đặc điểm
- Dùng tốt cho backtracking, tìm tất cả đường đi, kiểm tra tính liên thông.
- Nhưng đường đi không tối ưu
- Thứ tự neighbor ảnh hưởng mạnh đến kết quả
##### 3. Tìm kiếm đường đi ngắn nhất theo thuật toán Dijkstra
##### a. Phương pháp
Dijkstra dùng heap (priority queue) để luôn mở rộng node có chi phí từ start nhỏ nhất trước. 
- Lấy nút có chi phí nhỏ nhất, cập nhật chi phí đến các neighbor.
- Nếu tìm thấy chi phí nhỏ hơn, cập nhật dist và came_from.
- Khi đến goal, dừng và tạo đường đi từ came_from.

Nó đảm bảo tìm được đường đi ngắn nhất trên đồ thị có trọng số không âm.
##### b. Dữ liệu cần
- Heap (priority queue): `heap = [(0, start)]` lưu (chi phí, node) ưu tiên chi phí nhỏ.
- dist lưu chi phí nhỏ nhất từ start đến mỗi node.
- came_from để dựng đường đi.
- Graph phải có phương thức neighbors(node). Graph phải cung cấp trọng số cạnh (`w >= 0`).
##### c. Cách cài đặt
- Khởi tạo heap với start (`heap = [(0, start)]`), dist[start] = 0, came_from[start] = None. Đưa start vào hàng đợi `yield ("visit", start)`.
- Vòng lặp while heap:
    - Lấy node cur có chi phí nhỏ nhất từ heap (`d, cur = heapq.heappop(heap)`). Đánh dấu expand. Nếu node này không phải đường tốt nhất sẽ bị bỏ qua ở các lần sau.
    - Nếu node là goal, dừng vòng lặp.
    - Duyệt neighbor (nbr, w) của đỉnh hiện tại cur với w là trọng số của cạnh nối từ cur → nbr:
        - Tính chi phí mới khi đi qua node nbr: `nd = d + w`
        - `if nbr not in dist or nd < dist[nbr]` điều kiện kiểm tra xem có cần cập nhật chi phí ngắn nhất tới đỉnh nbr. Nếu neighbor chưa có trong dist hoặc `nd < dist[nbr]`, cập nhật:
            - Chi phí ngắn nhất hiện tại từ start → nbr là nd (`dist[nbr] = nd`).
            - Lưu vết đường đi `came_from[nbr] = cur`
            - Đưa node kề nbr vào heap, heap sẽ tự sắp xếp để node có chi phí nhỏ nhất được lấy ra trước (`heapq.heappush(heap, (nd, nbr))`), đánh dấu visit (`yield ("visit", nbr)`).
- Khi kết thúc, dựng đường đi bằng reconstruct_path(came_from, goal).
##### d. Đặc điểm
- Luôn tìm được đường đi ngắn nhất trên đồ thị trọng số không âm.
- Thời gian: O((V + E) log V).
##### 4. Tìm kiếm A* - A* search
##### a. Phương pháp
A* kết hợp chi phí thực tế từ start và heuristic ước lượng đến goal để mở rộng node một cách thông minh. 
- Dùng heap theo tổng chi phí ước lượng f = g + h, trong đó:
    - g: chi phí từ start tới nút hiện tại.
    - h: heuristic ước lượng chi phí từ nút tới goal.
- Mỗi lần mở rộng nút với f nhỏ nhất.
- Cập nhật g và came_from nếu tìm được đường tốt hơn.
- Khi đến goal, dựng đường đi từ came_from.
##### b. Dữ liệu cần
- Hàng đợi ưu tiên (heap): lưu các node theo thứ tự tăng dần của f = g + h.
- gscore: lưu chi phí nhỏ nhất hiện biết từ start đến mỗi node.
- Hàm heuristic h(n): dùng để ước lượng chi phí từ đỉnh hiện tại đến đích.
- came_from: lưu vết node cha để dựng lại đường đi cuối cùng.
- Graph có các đỉnh và các cạnh có trọng số không âm, với hàm neighbors(node) trả về các cặp (đỉnh kề, trọng số).
##### c. Cách cài đặt
- Khai báo hàm heuristic để ước lượng chi phí từ một đỉnh bất kỳ đến đỉnh đích. Heuristic là khoảng cách Manhattan: `h = lambda a, b: abs(a[0]-b[0]) + abs(a[1]-b[1])`
- Khởi tạo heap (priority queue) để lưu các đỉnh sẽ được mở rộng. Mỗi phần tử trong hàng đợi gồm ba thành phần (tổng chi phí ước lượng, chi phí thực từ start đến đỉnh hiện tại, đỉnh tương ứng). Khởi tạo ban đầu hàng đợi chỉ chứa đỉnh start (`heap = [(h(start, goal), 0, start)]`), `gscore = {start: 0}`, `came_from = {start: None}`. Đánh dấu `yield ("visit", start)`.
- Vòng lặp while heap:
    - Lấy ra node có giá trị f nhỏ nhất từ hàng đợi ưu tiên (`_, curg, cur = heapq.heappop(heap)`). Đánh dấu expand.
    - Nếu node đang xét là goal, vòng lặp dừng lại.
    - Duyệt neighbor (nbr, w) của đỉnh hiện tại cur với w là trọng số của cạnh nối từ cur → nbr:
        - Tính chi phí mới từ start đến nbr thông qua cur (`tentative_g = gscore[cur] + w`)
        - `if nbr not in gscore or tentative_g < gscore[nbr]` điều kiện cập nhật đường đi tốt hơn tới nbr. Nếu nbr chưa từng được thăm hoặc `tentative_g < gscore[nbr]` thì:
            - Cập nhật chi phí tốt nhất `gscore[nbr] = tentative_g`
            - Cập nhập giá trị f của A* `f = tentative_g + h(nbr, goal)` 
            - Lưu vết đường đi `came_from[nbr] = cur`.
            - Đưa đỉnh kề nbr vào hàng đợi ưu tiên (`heapq.heappush(heap, (f, tentative_g, nbr))`), đánh dấu visit (`yield ("visit", nbr)`).
- Khi kết thúc, dựng đường đi bằng reconstruct_path(came_from, goal).
##### d. Đặc điểm
- Tìm đường đi ngắn nhất sử dụng hàm heuristic để kết hợp chi phí thực và chi phí ước lượng, đảm bảo tối ưu và hiệu quả khi heuristic phù hợp.
- Thời gian chạy nhanh hơn BFS/Dijkstra, khi heuristic phù hợp, trong trường hợp xấu nhất là O(E log V).

```python
import random
import time
import heapq
from collections import deque, defaultdict

import matplotlib.pyplot as plt
import pandas as pd
```

##### 1. Sinh mê cung – `maze_generator`

```python
def generate_maze(rows, cols, extra_paths=40):
    maze = [[1 for _ in range(cols)] for _ in range(rows)]
    directions = [(0,1),(1,0),(0,-1),(-1,0)]

    def in_bounds(x, y):
        return 0 <= x < cols and 0 <= y < rows

    def carve(x, y):
        maze[y][x] = 0
        dirs = directions[:]
        random.shuffle(dirs)
        for dx, dy in dirs:
            nx, ny = x + dx*2, y + dy*2
            if in_bounds(nx, ny) and maze[ny][nx] == 1:
                maze[y + dy][x + dx] = 0
                carve(nx, ny)

    carve(0, 0)

    maze[rows-1][cols-1] = 0
    if maze[rows-2][cols-1] == 1 and maze[rows-1][cols-2] == 1:
        maze[rows-2][cols-1] = 0

    walls = [(y, x) for y in range(1, rows-1)
                     for x in range(1, cols-1) if maze[y][x] == 1]
    random.shuffle(walls)
    for y, x in walls[:extra_paths]:
        maze[y][x] = 0

    return maze
```

#### 2. Chuyển mê cung thành đồ thị – `graph`

```python
class Graph:
    def __init__(self):
        self.adj = defaultdict(list)

    def add_node(self, node):
        _ = self.adj[node]

    def add_edge(self, u, v, w=1):
        if u != v:
            self.adj[u].append((v, w))
            self.adj[v].append((u, w))

    def neighbors(self, node):
        return self.adj[node]

    def nodes(self):
        return list(self.adj.keys())


def grid_to_graph(maze):
    rows, cols = len(maze), len(maze[0])
    g = Graph()
    dirs = [(0,1),(1,0),(0,-1),(-1,0)]

    for r in range(rows):
        for c in range(cols):
            if maze[r][c] == 0:
                u = (r, c)
                g.add_node(u)
                for dr, dc in dirs:
                    nr, nc = r + dr, c + dc
                    if 0 <= nr < rows and 0 <= nc < cols and maze[nr][nc] == 0:
                        g.add_edge(u, (nr, nc), 1)
    return g
```

#### 3. Giải mê cung – `maze_solver`

```python
def reconstruct_path(came_from, goal):
    path = []
    cur = goal
    while cur is not None:
        path.append(cur)
        cur = came_from[cur]
    return path[::-1]


def bfs(graph, start, goal):
    q = deque([start])
    came = {start: None}
    yield ("visit", start)

    while q:
        cur = q.popleft()
        yield ("expand", cur)
        if cur == goal: break
        for nbr, _ in graph.neighbors(cur):
            if nbr not in came:
                came[nbr] = cur
                q.append(nbr)
                yield ("visit", nbr)

    yield ("path", reconstruct_path(came, goal) if goal in came else None)


def dfs(graph, start, goal):
    stack = [start]
    came = {start: None}
    yield ("visit", start)

    while stack:
        cur = stack.pop()
        yield ("expand", cur)
        if cur == goal: break
        for nbr, _ in graph.neighbors(cur):
            if nbr not in came:
                came[nbr] = cur
                stack.append(nbr)
                yield ("visit", nbr)

    yield ("path", reconstruct_path(came, goal) if goal in came else None)


def dijkstra(graph, start, goal):
    heap = [(0, start)]
    dist = {start: 0}
    came = {start: None}
    yield ("visit", start)

    while heap:
        d, cur = heapq.heappop(heap)
        yield ("expand", cur)
        if cur == goal: break
        for nbr, w in graph.neighbors(cur):
            nd = d + w
            if nbr not in dist or nd < dist[nbr]:
                dist[nbr] = nd
                came[nbr] = cur
                heapq.heappush(heap, (nd, nbr))
                yield ("visit", nbr)

    yield ("path", reconstruct_path(came, goal) if goal in came else None)


def astar(graph, start, goal):
    h = lambda a, b: abs(a[0]-b[0]) + abs(a[1]-b[1])
    heap = [(h(start, goal), 0, start)]
    gscore = {start: 0}
    came = {start: None}
    yield ("visit", start)

    while heap:
        _, g, cur = heapq.heappop(heap)
        yield ("expand", cur)
        if cur == goal: break
        for nbr, w in graph.neighbors(cur):
            ng = g + w
            if nbr not in gscore or ng < gscore[nbr]:
                gscore[nbr] = ng
                came[nbr] = cur
                heapq.heappush(heap, (ng + h(nbr, goal), ng, nbr))
                yield ("visit", nbr)

    yield ("path", reconstruct_path(came, goal) if goal in came else None)
```

#### 4. Kiểm thử tạo mê cung giống hiển thị của hệ thống, chạy và thu thập số liệu

```python
ROWS, COLS = 20, 25
maze = generate_maze(ROWS, COLS, extra_paths=40)
graph = grid_to_graph(maze)

start = (0, 0)
goal = (ROWS-1, COLS-1)

len(graph.nodes())
```

```python
def evaluate(name, solver):
    visited = set()
    expanded = 0
    max_memory = 0
    path = None

    t0 = time.perf_counter()

    for event, data in solver(graph, start, goal):
        if event == "visit":
            visited.add(data)
        elif event == "expand":
            expanded += 1
        elif event == "path":
            path = data

        max_memory = max(max_memory, len(visited))

    t1 = time.perf_counter()

    return {
        "Time (s)": t1 - t0,
        "Expanded nodes": expanded,
        "Memory (nodes)": max_memory,
        "Path length": len(path) if path else None,
        "Path": path
    }
```

```python
results = {
    "BFS": evaluate("BFS", bfs),
    "DFS": evaluate("DFS", dfs),
    "Dijkstra": evaluate("Dijkstra", dijkstra),
    "A*": evaluate("A*", astar)
}

df = pd.DataFrame(results).T
df
```

##### Phân tích 
##### 1. BFS (Breadth-First Search)
- **Thời gian**: BFS mở rộng tất cả các node theo mức từ start đến goal. Duyệt theo chiều rộng số node mở rộng nhiều nên thời gian khá cao.  
- **Node mở rộng**: BFS mở gần hết các node ở mức gần start trước khi tới goal, đảm bảo đường đi tối ưu.  
- **Bộ nhớ**: Lưu queue + visited, queue có thể chứa nhiều node cùng mức → tốn bộ nhớ.  
- **Độ dài đường đi**: BFS luôn tìm đường ngắn nhất vì mở rộng node theo mức (level). 

**Lý do kết quả**: BFS không dùng heuristic, mở rộng mọi khả năng theo level → đảm bảo đường ngắn nhất nhưng tốn nhiều node và bộ nhớ, thời gian vẫn còn cao.
##### 2. DFS (Depth-First Search)
- **Thời gian**: DFS đi sâu theo một nhánh tới goal. Trong một số maze này, DFS đi gần thẳng đến goal → thời gian rất thấp.  
- **Node mở rộng**: Chỉ mở một nhánh tại một thời điểm → ít node mở rộng.  
- **Bộ nhớ**: Stack lưu ít node → tiết kiệm bộ nhớ.  
- **Độ dài đường đi**: DFS không đảm bảo đường ngắn nhất, nên đường dài hơn BFS/A*/Dijkstra.  

**Lý do kết quả**: DFS nhanh và tiết kiệm memory nhưng không tối ưu đường đi. Nếu maze phức tạp hơn, DFS có thể đi vòng vèo nhiều hơn.
##### 3. Dijkstra
- **Thời gian**: Dijkstra dùng heap để chọn node chi phí thấp nhất. Với weight=1, hoạt động gần như BFS → thời gian tương đương.  
- **Node mở rộng**: Mở hầu hết node trước khi tới goal, giống BFS.  
- **Bộ nhớ**: Lưu heap + visited → tốn bộ nhớ.  
- **Độ dài đường đi**: Tối ưu nhờ luôn chọn node chi phí thấp nhất.  

**Lý do kết quả**: Khi trọng số =1, Dijkstra và BFS tương tự. Heap tạo overhead → thời gian hơi cao, nhưng đường đi tối ưu.
##### 4. A* (A star)
- **Thời gian**: A* dùng heuristic Manhattan hướng tới goal → mở ít node hơn BFS/Dijkstra → thời gian thấp hơn.  
- **Node mở rộng**: Mở ít node nhờ heuristic → tiết kiệm bộ nhớ và thời gian.  
- **Bộ nhớ**: Frontier nhỏ → bộ nhớ thấp, tương đương DFS.  
- **Độ dài đường đi**: Đường đi tối ưu nhờ heuristic dẫn hướng trực tiếp.  

**Lý do kết quả**: A* hiệu quả nhất, mở ít node, tối ưu đường đi, tiết kiệm memory và thời gian.
##### Kết luận tổng quát
- **DFS**: nhanh, ít memory, đường dài, không ổn định.  
- **BFS**: đường ngắn nhất, mở nhiều node, tốn memory.  
- **Dijkstra**: giống BFS khi weight=1, heap overhead.  
- **A***: hiệu quả nhất, mở ít node, đường tối ưu, tiết kiệm memory và thời gian.

#### Đánh giá kết quả thuật toán tìm đường 
Dựa trên dữ liệu chạy kiểm thử với nhiều kích thước maze và nhiều lần chạy:
##### DFS (Depth-First Search)
- **Ưu điểm**:  
    - Memory và node mở rộng ít hơn đáng kể so với BFS/Dijkstra/A*.  
- **Nhược điểm**:  
    - Đường đi **không tối ưu**, đặc biệt khi maze lớn → path length tăng mạnh 
    - Không đảm bảo đường ngắn nhất → không ổn định cho các ứng dụng cần tối ưu đường đi.
- **Đánh giá tổng quan**: DFS phù hợp khi ưu tiên tốc độ và tiết kiệm bộ nhớ, nhưng không tối ưu đường đi.
##### BFS (Breadth-First Search)
- **Ưu điểm**:  
    - Luôn tìm được **đường đi tối ưu**, path length ổn định.  
    - Đơn giản, dễ triển khai, đảm bảo tìm tất cả node theo mức từ start → goal.  
- **Nhược điểm**:  
    - Node mở rộng và Memory tăng rất nhanh khi maze lớn
    - Time tăng theo maze size, tốn tài nguyên cho maze lớn.  
- **Đánh giá tổng quan**: BFS phù hợp khi **maze nhỏ hoặc cần đường đi ngắn nhất**, nhưng kém hiệu quả cho maze lớn.
##### Dijkstra
- **Ưu điểm**:  
    - Đảm bảo đường đi tối ưu (path ngắn nhất theo tổng trọng số).  
    - Phù hợp khi maze có **trọng số khác nhau** giữa các ô.  
- **Nhược điểm**:  
    - Node mở rộng và Memory gần giống BFS → tốn tài nguyên nhiều.  
    - Time cao hơn BFS do **heap operations**, đặc biệt với maze lớn.  
- **Đánh giá tổng quan**: Dijkstra tốt cho maze có trọng số, nhưng khi trọng số = 1, A* hoặc BFS hiệu quả hơn.
##### A* (A star)
- **Ưu điểm**:  
    - Kết hợp **heuristic (Manhattan)** → mở ít node hơn BFS/Dijkstra → tiết kiệm Memory và giảm Time.  
    - Đường đi tối ưu, path length tương đương BFS/Dijkstra.  
    - Hiệu quả tốt với **maze lớn**, giảm node mở và Memory so với BFS/Dijkstra.  
- **Nhược điểm**:  
    - Cần tính heuristic, thuật toán phức tạp hơn BFS.  
    - Time vẫn tăng theo maze size, nhưng tổng thể cân bằng tốt giữa hiệu năng và resource.
- **Đánh giá tổng quan**: A* là lựa chọn tối ưu khi maze lớn cần đường đi ngắn, tiết kiệm Memory, giảm số node mở.
##### Tổng kết đánh giá
- Khi maze lớn, chỉ có A* và DFS chạy ổn định về memory; BFS/Dijkstra tốn nhiều resource.  
- DFS chỉ phù hợp khi **ưu tiên tốc độ và tiết kiệm memory**, không cần đường đi tối ưu.  
- A* là thuật toán **tốt nhất tổng thể**, cân bằng giữa Time, Memory, Node mở và Path length.
